```python
import pandas as pd
import numpy as np

scoring_with_arrangement = pd.read_csv("../data_processed/scoring_with_arrangement.csv", na_values=["NA"])

print("YEAR -", scoring_with_arrangement['YEAR'].isna().sum())
print("IP_ID -", scoring_with_arrangement['IP_ID'].isna().sum())
print("AR_ID -", scoring_with_arrangement['AR_ID'].isna().sum())
print("PD -", scoring_with_arrangement['PD'].isna().sum())
print("PD_POOL -", scoring_with_arrangement['PD_POOL'].isna().sum())
print("DFLT_FLAG -", scoring_with_arrangement['DFLT_FLAG'].isna().sum())
print("DEBT_RATIO -", scoring_with_arrangement['DEBT_RATIO'].isna().sum())
print("DPD -", scoring_with_arrangement['DPD'].isna().sum())
print("M_LAST_DPD -", scoring_with_arrangement['M_LAST_DPD'].isna().sum())
```

```python
df = scoring_with_arrangement[scoring_with_arrangement['PD'].notna()].copy()
group_missing = df[df['DPD'].isna()]
group_present = df[df['DPD'].notna()]

print("DPD missing n:", group_missing.shape[0])
print("DPD missing default_rate:", group_missing['DFLT_FLAG'].mean())
print("DPD missing avg_pd:", group_missing['PD'].mean())

print("DPD present n:", group_present.shape[0])
print("DPD present default_rate:", group_present['DFLT_FLAG'].mean())
print("DPD present avg_pd:", group_present['PD'].mean())
```

```python
df = scoring_with_arrangement[scoring_with_arrangement['PD'].notna()].copy()
df['dpd_missing'] = df['DPD'].isna()

print(df.groupby(['YEAR', 'dpd_missing'])['DFLT_FLAG'].mean())
print(df.groupby(['YEAR', 'dpd_missing'])['PD'].mean())
```
