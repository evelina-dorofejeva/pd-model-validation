```python
dat_arrangement['non_null_count'] = dat_arrangement.notna().sum(axis=1)
dat_arrangement.head()
```

```python
dat_arrangement_clean = (
    dat_arrangement
        .sort_values('non_null_count', ascending=False)
        .drop_duplicates(subset=['YEAR', 'AR_ID'])
)
print(dat_arrangement_clean.shape)
print(dat_arrangement_clean[['YEAR', 'AR_ID']].drop_duplicates().shape)
```

```python
scoring_with_arrangement = pd.merge(
    dat_scoring,
    dat_arrangement_clean,
    on=['YEAR', 'AR_ID'],
    how='left'
)
print(scoring_with_arrangement.shape)
print(scoring_with_arrangement.shape[0] == dat_scoring.shape[0])
scoring_with_arrangement.head()
```

```python
scoring_with_arrangement.to_csv("../data_processed/scoring_with_arrangement.csv", index=False)
```
